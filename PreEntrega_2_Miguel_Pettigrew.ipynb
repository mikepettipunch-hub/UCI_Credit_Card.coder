```python
duplicados = dataset[dataset.drop(columns=["ID"]).duplicated(keep=False)]
duplicados
```

```python
# además no tienen movimientos
montos = ["BILL_AMT1", "BILL_AMT2", "BILL_AMT3", "BILL_AMT4", "BILL_AMT5", "BILL_AMT6",
          "PAY_AMT1", "PAY_AMT2", "PAY_AMT3", "PAY_AMT4", "PAY_AMT5", "PAY_AMT6"]

sin_movimientos = duplicados[montos].sum(axis=1) == 0
print("Filas duplicadas:", len(duplicados))
print("Sin movimientos:", sin_movimientos.sum())
```

```python
# 4. función def 
# Función que clasifica el riesgo del cliente según sus meses de atraso
def clasificar_riesgo(meses_atraso):
    if meses_atraso <= 0:
        return "Al día"
    elif meses_atraso <= 2:
        return "Atraso leve"
    else:
        return "Atraso grave"

        # Prueba con un valor
print(clasificar_riesgo(3))   # Atraso grave
        
```

```python
# función lambda
# Calcula qué porcentaje de su límite de crédito está usando cada cliente
dataset["USO_LIMITE"] = dataset.apply(lambda fila: round(fila["BILL_AMT1"] / fila["LIMIT_BAL"] * 100, 1), axis=1)

dataset[["LIMIT_BAL", "BILL_AMT1", "USO_LIMITE"]].head()
```
