```python
# filas completamente iguales
df = df.drop_duplicates()
df = df.drop_duplicates(subset="ID", keep="first")
```

```python
# Subconjunto: clientes con límite bajo (50.000 o menos) que cayeron en mora
limite_bajo_mora = dataset[(dataset["LIMIT_BAL"] <= 50000) & (dataset["default.payment.next.month"] == 1)]

print("Clientes:", len(limite_bajo_mora))
```

```python
# códigos sin significado documentado -> los tratamos como faltantes (NaN)
df["EDUCATION"] = df["EDUCATION"].replace({0: np.nan, 5: np.nan, 6: np.nan})
df["MARRIAGE"] = df["MARRIAGE"].replace({0: np.nan})
```

```python
# imputación con la moda (el valor más frecuente)
for c in ["EDUCATION", "MARRIAGE"]:
    df[c] = df[c].fillna(df[c].mode()[0]).astype(int)
```

```python
# 1. Cantidad de NaN en cada columna
df[["EDUCATION", "MARRIAGE"]].isna().sum()
```

```python
# 3. Ver los valores que quedaron en cada columna y cuántas veces aparecen
print(df["EDUCATION"].value_counts().sort_index())
print(df["MARRIAGE"].value_counts().sort_index())
```

```python
# renombro la columna objetivo para que sea más corta
df = df.rename(columns={"default.payment.next.month": "default", "PAY_0": "PAY_1"})
```

```python
#AGRUPACIÓN 1:Pregunta: ¿qué porcentaje de clientes cae en default según su nivel educativo?

# diccionario: código numérico -> nombre del nivel (según la documentación del dataset)
niveles = {1: "Posgrado", 2: "Universidad", 3: "Secundario", 4: "Otros"}

# agrupación como antes
tasa_educ = df.groupby("EDUCATION")["default"].mean() * 100

# map() reemplaza cada número del índice por su nombre
tasa_educ.index = tasa_educ.index.map(niveles)

# nombre que aparece sobre el índice
tasa_educ.index.name = "Nivel educativo"

print(tasa_educ.round(1))
```

```python
# Agrupación 2
# Pregunta: ¿hay diferencia en la tasa de default entre hombres y mujeres?

# crea una columna con el nombre del sexo (1 = masculino, 2 = femenino, según la documentación)
df["SEXO"] = df["SEX"].map({1: "Masculino", 2: "Femenino"})

# agg calcula dos métricas: count = cantidad de clientes, mean = proporción de default
sexo = df.groupby("SEXO")["default"].agg(["count", "mean"])

# la proporción pasa a porcentaje, con un decimal
sexo["mean"] = (sexo["mean"] * 100).round(1)
print(sexo)
```

```python
# Agrupación 3
# Pregunta: ¿cómo cambia el riesgo de default según el límite de crédito?

# pd.cut arma rangos del límite de crédito; labels les pone nombre
df["rango_limite"] = pd.cut(df["LIMIT_BAL"],
                            bins=[0, 50000, 100000, 200000, 300000, 1000000],
                            labels=["hasta 50k", "50k-100k", "100k-200k", "200k-300k", "más de 300k"])

# misma estructura: cantidad de clientes y proporción de default por rango
limite = df.groupby("rango_limite", observed=True)["default"].agg(["count", "mean"])
limite["mean"] = (limite["mean"] * 100).round(1)
print(limite)
```
